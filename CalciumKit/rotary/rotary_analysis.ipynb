# Batch process raw rotary CSV files

This notebook processes every `.csv` file in aninput folder and saves the processed files into a subfolder called `processed`.

Expected columns in each raw CSV:
- `time_s`
- `position_deg`

```python
from pathlib import Path

from CalciumKit.behavior import process_rotary_csv
```

```python
# Define the folder containing raw rotary CSV files
input_dir = Path('/Users/suthardr/Desktop/Rotary/Recall2')

output_dir = input_dir / "processed"
output_dir.mkdir(parents=True, exist_ok=True)

processed_dir = output_dir

plot_dir = processed_dir / "plots"
plot_dir.mkdir(parents=True, exist_ok=True)

# Rotary wheel diameter in cm
D = 18

# Spline smoothing parameter assumption
sigma_deg = 0.1

print(f'Input folder: {input_dir}')
print(f'Output folder: {output_dir}')
```

## Process all CSV files

Uses `process_rotary_csv` from `CalciumKit.behavior`. Skips files already in the `processed` folder and avoids reprocessing files whose names begin with `processedRotary_`.

```python
csv_files = sorted(
    p for p in input_dir.glob("*.csv")
    if not p.name.startswith("processedRotary_")
)

print(f"Found {len(csv_files)} raw CSV file(s).")


results = []
errors = []

for csv_path in csv_files:
    try:
        result = process_rotary_csv(
            csv_path,
            output_dir,
            D=D,
            sigma_deg=sigma_deg
        )

        results.append(result)

        print(
            f"Processed: {csv_path.name} "
            f"-> {result['output_file']}"
        )

    except Exception as exc:
        errors.append({
            "file": csv_path.name,
            "error": str(exc)
        })

        print(
            f"ERROR processing {csv_path.name}: {exc}"
        )

summary = pd.DataFrame(results)
summary
```
