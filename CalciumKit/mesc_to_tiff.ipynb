# Femtonics mesc to tiff export

```python
from pathlib import Path

from CalciumKit.femtonics import extract_mesc_file
```

```python
MESC_FOLDER = Path(r"D:\ImagingData\Raw\Rachel")
OUT_ROOT    = Path(r"D:\ImagingData\Processed\Rachel")
```

```python
mesc_files = sorted(MESC_FOLDER.glob('*.mesc'))

if not mesc_files:
    print(f'No .mesc files found in {MESC_FOLDER}')
else:
    print(f'Found {len(mesc_files)} .mesc files:')
    for f in mesc_files:
        print(f'  {f.name}')

    failed = []
    for i, mesc_path in enumerate(mesc_files):
        print(f'\n[{i+1}/{len(mesc_files)}] {mesc_path.name}')
        try:
            extract_mesc_file(mesc_path, OUT_ROOT)
        except Exception as e:
            print(f'  ✗ FAILED: {e}')
            failed.append((mesc_path.name, str(e)))

    print(f'\n{"═" * 70}')
    print(f'Done.  {len(mesc_files) - len(failed)}/{len(mesc_files)} files processed successfully.')
    if failed:
        print('Failed:')
        for name, err in failed:
            print(f'  {name}: {err}')
```
